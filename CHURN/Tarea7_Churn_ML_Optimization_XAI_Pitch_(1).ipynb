2. **Validación:** 5-fold estratificada sobre el 80% de desarrollo. El 20% de prueba se usa una sola vez, en la evaluación final.
3. **ADASYN dentro de cada fold:** la síntesis usa solo las filas de entrenamiento de ese fold, nunca las de validación.
4. **Métricas:** Accuracy, Precision, Recall, F1, **F2** (β=2), **ROC-AUC**, **Brier Score** (calibración), latencia y tamaño del modelo.
5. **Optimización multiobjetivo NSGA-III (4 objetivos):** max F2, max ROC-AUC, min latencia (ms por 1,000 predicciones), min tamaño (KB serializado).
6. **XAI:** SHAP (TreeExplainer), correlación entre valor y SHAP, LIME y explicación contrafactual (búsqueda greedy sobre palancas accionables).

8. **Business Pitch:** ROI con supuestos hipotéticos, calculado en el simulador.

# 🧩 Parte 1. Preparación de datos, balanceo ADASYN y modelos predictivos

### Procedimientos:
1. **Carga y limpieza:** `TotalCharges` vacío se imputa con 0.0 (clientes nuevos, con tenure 0).
2. **Partición 80/20 estratificada:** el 20% de prueba (1,409 clientes) no se toca hasta la evaluación final.
3. **Validación 5-fold estratificada** sobre el 80% de desarrollo, con preprocesado y ADASYN ajustados dentro de cada fold.
4. **Modelos:** Regresión Logística, CatBoost y TabPFN, con las mismas particiones.
5. **Métricas:** Accuracy, Precision, Recall, F1, F2, ROC-AUC, Brier Score, latencia y tamaño.

```python
# 1. Carga del Dataset Telco Customer Churn
url = "https://raw.githubusercontent.com/IBM/telco-customer-churn-on-icp4d/master/data/Telco-Customer-Churn.csv"
if os.path.exists("Telco-Customer-Churn.csv"):
    df = pd.read_csv("Telco-Customer-Churn.csv")
else:
    df = pd.read_csv(url)
    df.to_csv("Telco-Customer-Churn.csv", index=False)

print(f"Dimensiones iniciales: {df.shape[0]} filas, {df.shape[1]} columnas")
display(df.head(3))
```

```python
# 2. Tratamiento de nulos, inconsistencias y target binario
df['TotalCharges'] = pd.to_numeric(df['TotalCharges'].replace(' ', np.nan), errors='coerce').fillna(0.0)
df['Churn_Binary'] = df['Churn'].apply(lambda x: 1 if x == 'Yes' else 0)

drop_cols = ['customerID', 'Churn', 'Churn_Binary']
feature_cols = [c for c in df.columns if c not in drop_cols]
X = df[feature_cols].copy()
y = df['Churn_Binary'].copy()

num_cols = ['tenure', 'MonthlyCharges', 'TotalCharges']
cat_cols = [c for c in feature_cols if c not in num_cols]

churn_dist = y.value_counts(normalize=True) * 100
print(f"Distribución original de Clases:")
print(f" - No Churn (0): {sum(y==0)} ({churn_dist[0]:.2f}%)")
print(f" - Churn (1) [Minoritaria]: {sum(y==1)} ({churn_dist[1]:.2f}%)")
```

```python
# 3. Partición estratificada 80/20, validación 5-fold y preprocesado
# El preprocesado se ajusta dentro de cada fold (ver make_pipeline_for en la celda siguiente).
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, stratify=y, random_state=SEED
)
CV = StratifiedKFold(n_splits=5, shuffle=True, random_state=SEED)

N_NUM = len(num_cols)  # numeric columns come first in the transformed matrix; the rest are one-hot (0/1)


def build_preprocessor():
    num_transformer = Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ])
    cat_transformer = Pipeline([
        ('imputer', SimpleImputer(strategy='most_frequent')),
        ('encoder', OneHotEncoder(drop='first', handle_unknown='ignore', sparse_output=False))
    ])
    return ColumnTransformer([
        ('num', num_transformer, num_cols),
        ('cat', cat_transformer, cat_cols)
    ])


print(f"Desarrollo (5-fold): {len(X_train)} filas | Prueba aislada: {len(X_test)} filas")
```

```python
# 4. ADASYN dentro de cada fold
# ADASYN interpola entre filas vecinas. Las columnas one-hot deben seguir siendo 0/1, así que tras el remuestreo se redondean.
from sklearn.base import BaseEstimator


class BinaryADASYN(BaseEstimator):
    def __init__(self, start_binary=N_NUM, random_state=SEED):
        self.start_binary = start_binary
        self.random_state = random_state

    def fit_resample(self, X, y):
        Xr, yr = ADASYN(random_state=self.random_state).fit_resample(X, y)
        Xr = np.asarray(Xr, dtype=float).copy()
        Xr[:, self.start_binary:] = np.clip(np.round(Xr[:, self.start_binary:]), 0, 1)
        return Xr, np.asarray(yr)


def make_pipeline_for(clf):
    """Preprocessing and ADASYN are fitted only on the rows they receive (the training rows of each fold)."""
    return ImbPipeline([
        ('prep', build_preprocessor()),
        ('adasyn', BinaryADASYN()),
        ('clf', clf),
    ])


# Illustration only: class counts before and after ADASYN on the development rows.
# The evaluation never uses this matrix; each fold resamples its own training rows.
prep_demo = build_preprocessor().fit(X_train)
X_demo, y_demo = BinaryADASYN().fit_resample(prep_demo.transform(X_train), y_train)

fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.countplot(x=y_train, ax=axes[0], hue=y_train, palette=['#3B82F6', '#EF4444'], legend=False)
axes[0].set_title(f"Antes de ADASYN (desarrollo)\nNo Churn: {sum(y_train==0)} | Churn: {sum(y_train==1)}")
axes[0].set_xticklabels(['No Churn (0)', 'Churn (1)'])
sns.countplot(x=y_demo, ax=axes[1], hue=y_demo, palette=['#3B82F6', '#F97316'], legend=False)
axes[1].set_title(f"Después de ADASYN (ilustración)\nNo Churn: {sum(y_demo==0)} | Churn: {sum(y_demo==1)}")
axes[1].set_xticklabels(['No Churn (0)', 'Churn sintético (1)'])
plt.tight_layout()
plt.show()
print(f"Columnas: {N_NUM} numéricas + {X_demo.shape[1] - N_NUM} binarias one-hot")
```

```python
# 5. Modelos base con 5-fold CV sobre el 80% de desarrollo (mismas particiones para todos)
def benchmark_model(model, X_sample, repeats=5):
    """Latency in ms per 1,000 predictions (median of repeats after one warm-up) and size in KB."""
    X_k = X_sample.iloc[:1000] if len(X_sample) >= 1000 else X_sample
    model.predict_proba(X_k)  # warm-up
    times = []
    for _ in range(repeats):
        t0 = time.perf_counter()
        model.predict_proba(X_k)
        times.append(time.perf_counter() - t0)
    lat_ms_1k = float(np.median(times)) * 1000 * (1000 / len(X_k))
    tmp = os.path.join(tempfile.gettempdir(), f"m_bench_{time.time_ns()}.joblib")
    joblib.dump(model, tmp, compress=3)
    sz_kb = os.path.getsize(tmp) / 1024.0
    os.remove(tmp)
    return lat_ms_1k, sz_kb


import torch
tabpfn_device = "cuda" if torch.cuda.is_available() else "cpu"

models = {
    "Logistic Regression": LogisticRegression(max_iter=1000, random_state=SEED),
    "Random Forest": RandomForestClassifier(n_estimators=100, max_depth=8, class_weight='balanced', random_state=SEED),
    "CatBoost": CatBoostClassifier(
        iterations=150, depth=5, learning_rate=0.08, verbose=0,
        random_seed=SEED, allow_writing_files=False, train_dir=tempfile.gettempdir()
    ),
}
try:
    models["TabPFN"] = TabPFNClassifier(device=tabpfn_device, random_state=SEED)
except Exception as e:
    print(f"⚠️ TabPFN no pudo inicializarse ({e}). Se continúa con Logistic Regression y CatBoost.")


def cv_scores(model):
    rows = []
    for tr, va in CV.split(X_train, y_train):
        pipe = make_pipeline_for(clone(model)).fit(X_train.iloc[tr], y_train.iloc[tr])
        proba = pipe.predict_proba(X_train.iloc[va])[:, 1]
        yv = y_train.iloc[va]
        pred = (proba >= 0.5).astype(int)
        rows.append({
            'Accuracy': accuracy_score(yv, pred),
            'Precision': precision_score(yv, pred, zero_division=0),
            'Recall': recall_score(yv, pred),
            'F1-Score': f1_score(yv, pred),
            'F2-Score': fbeta_score(yv, pred, beta=2),
            'ROC-AUC': roc_auc_score(yv, proba),
            'Brier Score': brier_score_loss(yv, proba),
        })
    return pd.DataFrame(rows)


baseline_results, baseline_folds = [], {}
for name, clf in models.items():
    try:
        fold_df = cv_scores(clf)
        baseline_folds[name] = fold_df
        full = make_pipeline_for(clone(clf)).fit(X_train, y_train)
        lat_ms, sz_kb = benchmark_model(full, X_train)
        row = {'Modelo': name}
        row.update({k: round(v, 4) for k, v in fold_df.mean().items()})
        row.update({'Latencia (ms/1k)': round(lat_ms, 2), 'Tamaño (KB)': round(sz_kb, 1)})
        baseline_results.append(row)
    except Exception as e:
        print(f"⚠️ No se pudo evaluar {name} en 5-fold CV ({e}). Se omite de la comparativa.")

baseline_table = pd.DataFrame(baseline_results).set_index('Modelo')
print("Media de 5 folds (desarrollo):")
display(baseline_table)
print("Desviación estándar entre folds:")
display(pd.DataFrame({k: v.std().round(4) for k, v in baseline_folds.items()}).T)
```

# ⚙️ Parte 2. Optimización de Hiperparámetros (Tradicionales vs Avanzadas) y Formulación Multiobjetivo

En esta sección comparamos rigurosamente cuatro estrategias de optimización sobre el mismo modelo base (**CatBoostClassifier**) y con el mismo protocolo de validación cruzada (**Stratified 5-Fold CV** sobre el 80% de desarrollo):

1. **Grid Search (Exhaustivo Tradicional):** Evalúa todas las combinaciones posibles en una rejilla cartesiana discreta. Determinista pero ineficiente dimensionalmente $\mathcal{O}(k^d)$.
2. **Random Search (Muestreo Aleatorio):** Muestrea configuraciones al azar desde distribuciones continuas y discretas (Bergstra & Bengio, 2012).
3. **Optimización Bayesiana (TPE con Optuna):** Modela la distribución condicional $p(\theta \mid y)$ guiando adaptativamente la búsqueda hacia zonas óptimas de $F_2$.
4. **Optimización Multiobjetivo (NSGA-III con Optuna):** Algoritmo evolutivo que busca simultáneamente optimizar precisión predictiva ($F_2$, ROC-AUC) y eficiencia en producción (Latencia, Tamaño).

### Espacios de Búsqueda de Hiperparámetros:
| Hiperparámetro | Tipo | Espacio de Búsqueda | Descripción |
| :--- | :---: | :---: | :--- |
| `depth` | Entero | [3, 7] | Profundidad de los árboles simétricos de CatBoost. |
| `iterations` | Entero | [50, 200] (step=25) | Número de árboles en el ensamble boosting. |
| `learning_rate` | Continuo | [0.03, 0.20] (log-scale) | Tasa de aprendizaje / gradiente. |
| `l2_leaf_reg` | Continuo | [1.0, 10.0] | Regularización $L_2$ de las hojas para prevenir sobreajuste. |


```python
# 1. Ejecución de las 4 Estrategias de Optimización sobre CatBoost (5-Fold CV)
# Evaluadas con las mismas particiones de desarrollo (80%) y make_pipeline_for
print('🚀 Iniciando benchmark real de las 4 estrategias de optimización...')

# --- A. Grid Search (Exhaustivo Tradicional) ---
grid_param_grid = {
    'depth': [3, 5],
    'iterations': [50, 100],
    'learning_rate': [0.05, 0.1]
}
grid_configs = [
    {'depth': d, 'iterations': it, 'learning_rate': lr, 'l2_leaf_reg': 5.0}
    for d in grid_param_grid['depth']
    for it in grid_param_grid['iterations']
    for lr in grid_param_grid['learning_rate']
]

print(f' • [1/4] Grid Search: Evaluando {len(grid_configs)} combinaciones discretas en 5-fold CV...')
t0_grid = time.time()
grid_best_score = -1.0
grid_best_params = None
grid_best_metrics = None

for cfg in grid_configs:
    clf = CatBoostClassifier(**cfg, verbose=0, random_seed=SEED, allow_writing_files=False, train_dir=tempfile.gettempdir(), thread_count=-1)
    f2_list, roc_list, rec_list, prec_list = [], [], [], []
    for tr, va in CV.split(X_train, y_train):
        pipe = make_pipeline_for(clone(clf)).fit(X_train.iloc[tr], y_train.iloc[tr])
        proba = pipe.predict_proba(X_train.iloc[va])[:, 1]
        yv = y_train.iloc[va]
        pred = (proba >= 0.5).astype(int)
        f2_list.append(fbeta_score(yv, pred, beta=2))
        roc_list.append(roc_auc_score(yv, proba))
        rec_list.append(recall_score(yv, pred))
        prec_list.append(precision_score(yv, pred, zero_division=0))
    mean_f2 = float(np.mean(f2_list))
    if mean_f2 > grid_best_score:
        grid_best_score = mean_f2
        grid_best_params = cfg
        grid_best_metrics = {
            'F2_Score': round(mean_f2, 4),
            'ROC_AUC': round(float(np.mean(roc_list)), 4),
            'Recall': round(float(np.mean(rec_list)), 4),
            'Precision': round(float(np.mean(prec_list)), 4)
        }

t_grid = time.time() - t0_grid
pipe_grid = make_pipeline_for(CatBoostClassifier(**grid_best_params, verbose=0, random_seed=SEED, allow_writing_files=False, train_dir=tempfile.gettempdir(), thread_count=-1)).fit(X_train, y_train)
lat_grid, sz_grid = benchmark_model(pipe_grid, X_train)
print(f'   ✅ Grid Search finalizado en {t_grid:.1f}s | Mejor F2: {grid_best_score:.4f}')

# --- B. Random Search (Muestreo Aleatorio) ---
print(' • [2/4] Random Search: Evaluando 10 configuraciones aleatorias...')
t0_rnd = time.time()
rng = np.random.RandomState(SEED)
rnd_configs = [
    {
        'depth': int(rng.choice([3, 4, 5, 6])),
        'iterations': int(rng.choice([50, 75, 100, 125, 150])),
        'learning_rate': float(np.round(rng.uniform(0.04, 0.15), 4)),
        'l2_leaf_reg': float(np.round(rng.uniform(2.0, 8.0), 2))
    }
    for _ in range(10)
]
rnd_best_score = -1.0
rnd_best_params = None
rnd_best_metrics = None

for cfg in rnd_configs:
    clf = CatBoostClassifier(**cfg, verbose=0, random_seed=SEED, allow_writing_files=False, train_dir=tempfile.gettempdir(), thread_count=-1)
    f2_list, roc_list, rec_list, prec_list = [], [], [], []
    for tr, va in CV.split(X_train, y_train):
        pipe = make_pipeline_for(clone(clf)).fit(X_train.iloc[tr], y_train.iloc[tr])
        proba = pipe.predict_proba(X_train.iloc[va])[:, 1]
        yv = y_train.iloc[va]
        pred = (proba >= 0.5).astype(int)
        f2_list.append(fbeta_score(yv, pred, beta=2))
        roc_list.append(roc_auc_score(yv, proba))
        rec_list.append(recall_score(yv, pred))
        prec_list.append(precision_score(yv, pred, zero_division=0))
    mean_f2 = float(np.mean(f2_list))
    if mean_f2 > rnd_best_score:
        rnd_best_score = mean_f2
        rnd_best_params = cfg
        rnd_best_metrics = {
            'F2_Score': round(mean_f2, 4),
            'ROC_AUC': round(float(np.mean(roc_list)), 4),
            'Recall': round(float(np.mean(rec_list)), 4),
            'Precision': round(float(np.mean(prec_list)), 4)
        }

t_rnd = time.time() - t0_rnd
pipe_rnd = make_pipeline_for(CatBoostClassifier(**rnd_best_params, verbose=0, random_seed=SEED, allow_writing_files=False, train_dir=tempfile.gettempdir(), thread_count=-1)).fit(X_train, y_train)
lat_rnd, sz_rnd = benchmark_model(pipe_rnd, X_train)
print(f'   ✅ Random Search finalizado en {t_rnd:.1f}s | Mejor F2: {rnd_best_score:.4f}')

# --- C. Optimización Bayesiana (TPE con Optuna) ---
print(' • [3/4] Optimización Bayesiana TPE: Evaluando 12 trials para maximizar F2...')
def objective_tpe(trial):
    params = dict(
        depth=trial.suggest_int('depth', 3, 6),
        iterations=trial.suggest_int('iterations', 50, 150, step=25),
        learning_rate=trial.suggest_float('learning_rate', 0.04, 0.15, log=True),
        l2_leaf_reg=trial.suggest_float('l2_leaf_reg', 2.0, 8.0),
    )
    clf = CatBoostClassifier(**params, verbose=0, random_seed=SEED, allow_writing_files=False, train_dir=tempfile.gettempdir(), thread_count=-1)
    f2_list = []
    for tr, va in CV.split(X_train, y_train):
        pipe = make_pipeline_for(clone(clf)).fit(X_train.iloc[tr], y_train.iloc[tr])
        proba = pipe.predict_proba(X_train.iloc[va])[:, 1]
        yv = y_train.iloc[va]
        f2_list.append(fbeta_score(yv, (proba >= 0.5).astype(int), beta=2))
    return float(np.mean(f2_list))

t0_tpe = time.time()
study_tpe = optuna.create_study(direction='maximize', sampler=optuna.samplers.TPESampler(seed=SEED))
study_tpe.optimize(objective_tpe, n_trials=12)
t_tpe = time.time() - t0_tpe
best_tpe_params = study_tpe.best_params

clf_tpe = CatBoostClassifier(**best_tpe_params, verbose=0, random_seed=SEED, allow_writing_files=False, train_dir=tempfile.gettempdir(), thread_count=-1)
f2_list, roc_list, rec_list, prec_list = [], [], [], []
for tr, va in CV.split(X_train, y_train):
    pipe = make_pipeline_for(clone(clf_tpe)).fit(X_train.iloc[tr], y_train.iloc[tr])
    proba = pipe.predict_proba(X_train.iloc[va])[:, 1]
    yv = y_train.iloc[va]
    pred = (proba >= 0.5).astype(int)
    f2_list.append(fbeta_score(yv, pred, beta=2))
    roc_list.append(roc_auc_score(yv, proba))
    rec_list.append(recall_score(yv, pred))
    prec_list.append(precision_score(yv, pred, zero_division=0))

tpe_best_metrics = {
    'F2_Score': round(float(np.mean(f2_list)), 4),
    'ROC_AUC': round(float(np.mean(roc_list)), 4),
    'Recall': round(float(np.mean(rec_list)), 4),
    'Precision': round(float(np.mean(prec_list)), 4)
}
pipe_tpe = make_pipeline_for(CatBoostClassifier(**best_tpe_params, verbose=0, random_seed=SEED, allow_writing_files=False, train_dir=tempfile.gettempdir(), thread_count=-1)).fit(X_train, y_train)
lat_tpe, sz_tpe = benchmark_model(pipe_tpe, X_train)
print(f'   ✅ Bayesiano TPE finalizado en {t_tpe:.1f}s | Mejor F2: {study_tpe.best_value:.4f}')

# --- D. Optimización Multiobjetivo (NSGA-III - 4 objetivos) ---
print(' • [4/4] NSGA-III Multiobjetivo: Evaluando 25 trials sobre [Max F2, Min Latencia, Min Tamaño, Max AUC]...')
def objective_nsgaiii(trial):
    params = dict(
        depth=trial.suggest_int('depth', 3, 6),
        iterations=trial.suggest_int('iterations', 50, 150, step=25),
        learning_rate=trial.suggest_float('learning_rate', 0.04, 0.15, log=True),
        l2_leaf_reg=trial.suggest_float('l2_leaf_reg', 2.0, 8.0),
    )
    def make_model():
        return make_pipeline_for(CatBoostClassifier(
            **params, verbose=0, random_seed=SEED, allow_writing_files=False,
            train_dir=tempfile.gettempdir(), thread_count=-1))

    f2_list, roc_list = [], []
    for tr, va in CV.split(X_train, y_train):
        pipe = make_model().fit(X_train.iloc[tr], y_train.iloc[tr])
        proba = pipe.predict_proba(X_train.iloc[va])[:, 1]
        yv = y_train.iloc[va]
        f2_list.append(fbeta_score(yv, (proba >= 0.5).astype(int), beta=2))
        roc_list.append(roc_auc_score(yv, proba))

    final = make_model().fit(X_train, y_train)
    lat_ms, sz_kb = benchmark_model(final, X_train)
    return float(np.mean(f2_list)), lat_ms, sz_kb, float(np.mean(roc_list))

t0_nsga = time.time()
sampler_nsga = NSGAIIISampler(population_size=10, seed=SEED)
study_nsga = optuna.create_study(
    directions=['maximize', 'minimize', 'minimize', 'maximize'],
    sampler=sampler_nsga
)
study_nsga.optimize(objective_nsgaiii, n_trials=25)
t_nsga = time.time() - t0_nsga
print(f'   ✅ NSGA-III finalizado en {t_nsga:.1f}s con {len(study_nsga.trials)} evaluaciones.')

# Consolidar lista de diccionarios optimization_comparison
pareto_trials = study_nsga.best_trials
best_nsga_t = max(pareto_trials, key=lambda t: t.values[0] * 0.5 + t.values[3] * 0.5 - t.values[1]*0.01)
champion_config = best_nsga_t.params

clf_nsga = CatBoostClassifier(**champion_config, verbose=0, random_seed=SEED, allow_writing_files=False, train_dir=tempfile.gettempdir(), thread_count=-1)
f2_l, roc_l, rec_l, prec_l = [], [], [], []
for tr, va in CV.split(X_train, y_train):
    pipe = make_pipeline_for(clone(clf_nsga)).fit(X_train.iloc[tr], y_train.iloc[tr])
    proba = pipe.predict_proba(X_train.iloc[va])[:, 1]
    yv = y_train.iloc[va]
    pred = (proba >= 0.5).astype(int)
    f2_l.append(fbeta_score(yv, pred, beta=2))
    roc_l.append(roc_auc_score(yv, proba))
    rec_l.append(recall_score(yv, pred))
    prec_l.append(precision_score(yv, pred, zero_division=0))

nsga_metrics = {
    'F2_Score': round(float(np.mean(f2_l)), 4),
    'ROC_AUC': round(float(np.mean(roc_l)), 4),
    'Recall': round(float(np.mean(rec_l)), 4),
    'Precision': round(float(np.mean(prec_l)), 4)
}

optimization_comparison = [
    {
        'Estrategia': 'Grid Search',
        'Tipo': 'Tradicional Exhaustivo',
        'Objetivos': 'Mono-objetivo (F2)',
        'N_Evaluaciones': len(grid_configs),
        'Tiempo_s': round(t_grid, 2),
        'F2_Score': grid_best_metrics['F2_Score'],
        'ROC_AUC': grid_best_metrics['ROC_AUC'],
        'Recall': grid_best_metrics['Recall'],
        'Precision': grid_best_metrics['Precision'],
        'Latencia_ms': round(lat_grid, 2),
        'Tamano_KB': round(sz_grid, 1),
        'Ventajas': 'Garantiza cobertura uniforme en malla discreta; determinista.',
        'Limitaciones': 'Explosión combinatoria O(k^d); ineficiente en parámetros continuos.'
    },
    {
        'Estrategia': 'Random Search',
        'Tipo': 'Muestreo Aleatorio',
        'Objetivos': 'Mono-objetivo (F2)',
        'N_Evaluaciones': len(rnd_configs),
        'Tiempo_s': round(t_rnd, 2),
        'F2_Score': rnd_best_metrics['F2_Score'],
        'ROC_AUC': rnd_best_metrics['ROC_AUC'],
        'Recall': rnd_best_metrics['Recall'],
        'Precision': rnd_best_metrics['Precision'],
        'Latencia_ms': round(lat_rnd, 2),
        'Tamano_KB': round(sz_rnd, 1),
        'Ventajas': 'Explora eficazmente espacios de búsqueda continuos y multidimensionales.',
        'Limitaciones': 'Muestreo ciego: no acumula conocimiento de las iteraciones previas.'
    },
    {
        'Estrategia': 'Bayesiano (TPE)',
        'Tipo': 'Avanzado Probabilístico',
        'Objetivos': 'Mono-objetivo (F2)',
        'N_Evaluaciones': len(study_tpe.trials),
        'Tiempo_s': round(t_tpe, 2),
        'F2_Score': tpe_best_metrics['F2_Score'],
        'ROC_AUC': tpe_best_metrics['ROC_AUC'],
        'Recall': tpe_best_metrics['Recall'],
        'Precision': tpe_best_metrics['Precision'],
        'Latencia_ms': round(lat_tpe, 2),
        'Tamano_KB': round(sz_tpe, 1),
        'Ventajas': 'Modela la distribución p(theta|y) para converger a óptimos con pocos ensayos.',
        'Limitaciones': 'Mono-objetivo: optimiza F2 ignorando la latencia y el peso del modelo.'
    },
    {
        'Estrategia': 'Multiobjetivo (NSGA-III)',
        'Tipo': 'Evolutivo Multiobjetivo',
        'Objetivos': '4 Objetivos: Max F2, Max AUC, Min Latencia, Min Tamaño',
        'N_Evaluaciones': len(study_nsga.trials),
        'Tiempo_s': round(t_nsga, 2),
        'F2_Score': nsga_metrics['F2_Score'],
        'ROC_AUC': nsga_metrics['ROC_AUC'],
        'Recall': nsga_metrics['Recall'],
        'Precision': nsga_metrics['Precision'],
        'Latencia_ms': round(best_nsga_t.values[1], 2),
        'Tamano_KB': round(best_nsga_t.values[2], 1),
        'Ventajas': 'Encuentra el Frente de Pareto equilibrando efectividad predictiva y costo operacional.',
        'Limitaciones': 'Requiere mayor presupuesto computacional para poblar el frente multidimensional.'
    }
]
print('📊 optimization_comparison consolidada con éxito.')
```

```python
# 3. Cálculo de LIME, Casos Individuales (Cliente A y B) y Comparativa con Random Forest
lime_explainer = lime.lime_tabular.LimeTabularExplainer(
    training_data=X_train_proc,
    feature_names=feature_names,
    class_names=['Retiene (0)', 'Abandona (1)'],
    mode='classification',
    random_state=SEED
)

# Identificación de los 2 Clientes de prueba (Alto Riesgo y Bajo Riesgo)
high_risk_idx = int(np.argmax(y_prob_champ))
low_risk_idx = int(np.argmin(y_prob_champ))

def extract_client_xai(idx, name_label):
    p = float(y_prob_champ[idx])
    raw = X_test.iloc[idx].to_dict()
    x_row = X_test_proc[idx]
    s_vals = explainer_shap.shap_values(x_row.reshape(1, -1))[0]
    top_s_idx = np.argsort(np.abs(s_vals))[::-1][:6]
    shap_factors = [
        {
            'Feature': feature_names[i],
            'Valor_Original': str(raw.get(feature_names[i], round(float(x_row[i]), 2))),
            'SHAP_Impacto': round(float(s_vals[i]), 4),
            'Efecto': 'Aumenta Abandono (+)' if s_vals[i] > 0 else 'Disminuye Abandono (-)'
        }
        for i in top_s_idx
    ]
    exp_l = lime_explainer.explain_instance(x_row, champion_model.predict_proba, num_features=6)
    lime_factors = [
        {
            'Regla': r,
            'Peso_LIME': round(w, 4),
            'Efecto': 'Aumenta Abandono (+)' if w > 0 else 'Disminuye Abandono (-)'
        }
        for r, w in exp_l.as_list()
    ]
    return {
        'id': f"Cliente_{name_label}",
        'probabilidad_abandono': round(p, 4),
        'clasificacion_real': int(y_test.iloc[idx]),
        'perfil_resumen': {
            'Contract': raw.get('Contract'),
            'tenure': int(raw.get('tenure')),
            'MonthlyCharges': float(raw.get('MonthlyCharges')),
            'InternetService': raw.get('InternetService'),
            'TechSupport': raw.get('TechSupport'),
            'PaymentMethod': raw.get('PaymentMethod')
        },
        'shap_factores': shap_factors,
        'lime_factores': lime_factors
    }

individual_xai_cases = {
    'cliente_a_alto_riesgo': extract_client_xai(high_risk_idx, 'A_Alto_Riesgo'),
    'cliente_b_bajo_riesgo': extract_client_xai(low_risk_idx, 'B_Bajo_Riesgo')
}

# Entrenamiento de Random Forest para Feature Importance (MDI)
rf_clf = RandomForestClassifier(n_estimators=100, max_depth=8, class_weight='balanced', random_state=SEED).fit(X_train_proc, y_train)
rf_mdi = rf_clf.feature_importances_

top_shap_idx = np.argsort(mean_abs_shap)[::-1][:10]
rf_vs_shap_importance = []
for idx in top_shap_idx:
    rf_vs_shap_importance.append({
        'Feature': feature_names[idx],
        'SHAP_Importance': round(float(mean_abs_shap[idx]), 4),
        'RF_MDI_Importance': round(float(rf_mdi[idx]), 4)
    })

print('✅ Casos individuales de XAI y comparativa Random Forest calculados con éxito.')
```

### 🎯 4. Análisis Contrafactual Prescriptivo (Búsqueda Greedy sobre Palancas Accionables)

Para un cliente con riesgo $\ge 0.50$, la función prueba palancas comerciales en un orden fijo. Aplica cada una si baja el riesgo. Se detiene al llegar a $\le 0.30$ o al agotar las palancas.

```python
# Motor contrafactual prescriptivo
def find_counterfactual(client_dict, model, preprocessor, target_prob=0.30):
    client_df = pd.DataFrame([client_dict])
    init_p = float(model.predict_proba(preprocessor.transform(client_df))[0, 1])
    if init_p <= target_prob:
        return {'status': 'Seguro', 'prob_inicial': init_p, 'prob_final': init_p, 'pasos': []}
    cf_candidate = client_dict.copy()
    steps = []
    levers = [
        ('Contract', 'Two year', 'Migrar a contrato por 2 años'),
        ('TechSupport', 'Yes', 'Bonificar Soporte Técnico prioritario'),
        ('OnlineSecurity', 'Yes', 'Activar suite de Seguridad en Línea'),
        ('PaymentMethod', 'Bank transfer (automatic)', 'Descuento por domiciliar pago a Débito Bancario'),
        ('MonthlyCharges', max(20.0, float(client_dict.get('MonthlyCharges', 70)) - 15.0), 'Bono promocional de -$15/mes'),
        ('Contract', 'One year', 'Migrar a contrato anual')
    ]
    curr_p = init_p
    for feat, val, desc in levers:
        if cf_candidate[feat] == val:
            continue
        old_val = cf_candidate[feat]
        trial = cf_candidate.copy()
        trial[feat] = val
        new_p = float(model.predict_proba(preprocessor.transform(pd.DataFrame([trial])))[0, 1])
        if new_p >= curr_p:
            continue
        drop = curr_p - new_p
        cf_candidate = trial
        steps.append({
            'Palanca Prescriptiva': desc,
            'Variable': feat,
            'Antes': old_val,
            'Después': val,
            'Probabilidad Reducida': f"{new_p:.1%}",
            'Impacto': f"-{drop:.1%}"
        })
        curr_p = new_p
        if curr_p <= target_prob:
            break
    status = 'Rescatado' if curr_p <= target_prob else 'No alcanza el umbral'
    return {'status': status, 'prob_inicial': init_p, 'prob_final': curr_p, 'pasos': steps}

cliente_riesgo = X_test.iloc[high_risk_idx].to_dict()
cf_res = find_counterfactual(cliente_riesgo, champion_model, preprocessor)
print(f"Probabilidad inicial: {cf_res['prob_inicial']:.1%} ──> Contrafactual: {cf_res['prob_final']:.1%} ({cf_res['status']})")
display(pd.DataFrame(cf_res['pasos']))
```

### 📚 Análisis Comparativo y Respuestas Oficiales a Preguntas de la Rúbrica (XAI)

#### 1. ¿Qué similitudes y diferencias encontró entre SHAP y LIME?
- **Similitudes:** Ambos son métodos de explicabilidad local agnósticos al modelo (model-agnostic) que descomponen la probabilidad predicha asignando pesos de contribución a cada variable para un cliente individual.
- **Diferencias Fundamentales:**
  - **SHAP (SHapley Additive exPlanations):** Se fundamenta en la teoría de juegos cooperativos (Shapley values). Garantiza cuatro propiedades axiomáticas esenciales: *Eficiencia* (suma aditiva exacta a la predicción), *Simetría*, *Dummy* y *Consistencia*. Su solución matemática es única y determinista.
  - **LIME (Local Interpretable Model-agnostic Explanations):** Entrena un modelo lineal interpretable subrogado local ponderado por una función kernel exponencial sobre perturbaciones generadas en el vecindario del cliente. No tiene garantías axiomáticas y puede variar estocásticamente entre corridas según el muestreo aleatorio.

#### 2. ¿Qué características resultaron relevantes para las predicciones?
- **Tipo de Contrato (`Contract`):** `Month-to-month` es el mayor detonante del riesgo de fuga (+SHAP), mientras que `Two year` brinda la mayor retención estructural (-SHAP).
- **Antigüedad del Cliente (`tenure`):** Clientes noveles (<12 meses) concentran el riesgo crítico, estabilizándose a partir de los 48 meses.
- **Servicio de Internet y Soporte:** `Fiber optic` sin `TechSupport` dispara las deserciones debido a insatisfacción con el servicio o alta tarifa sin soporte VIP.
- **Método de Pago (`PaymentMethod`):** `Electronic check` presenta alta fricción operativa frente a las domiciliaciones bancarias automáticas.

#### 3. ¿Cómo explicaría una predicción a una persona sin conocimientos técnicos?
> *«Imagine que la decisión del cliente es una balanza. Cada servicio contratado o condición del cliente coloca una pesa en uno de los dos platillos. Por ejemplo, tener un contrato mes a mes y pagar por cheque electrónico colocan pesas pesadas en el platillo de 'Se marchará'. En cambio, llevar 4 años con la empresa y contar con soporte técnico prioritario colocan pesas en el platillo de 'Se quedará'. La predicción final no es magia: es simplemente hacia qué lado se inclina la balanza con todos sus servicios combinados.»*

#### 4. ¿Qué limitaciones presentan ambas técnicas?
- **Limitaciones de SHAP:** Alto costo computacional (evaluación combinatoria $2^F$); asunción de independencia condicional entre variables correlacionadas en implementaciones rápidas.
- **Limitaciones de LIME:** Inestabilidad en las explicaciones locales debido al muestreo aleatorio; sensibilidad crítica al ancho de banda del kernel; e incapacidad de aproximar bien fronteras altamente no lineales mediante un modelo lineal simple.

#### 5. Comparativa con Feature Importance de Random Forest (MDI vs SHAP)
- El **Mean Decrease in Impurity (MDI / Gini Importance)** de Random Forest mide cuánta impureza reduce una variable en los nodos de entrenamiento. Presenta un **sesgo sistemático hacia variables continuas y de alta cardinalidad** (como `MonthlyCharges` o `TotalCharges`), sobreestimando su peso real.
- **SHAP** mide el impacto real directo en la probabilidad marginal predicha en el conjunto de prueba aislado, posicionando a variables estructurales discretas (como `Contract_Two year` y `InternetService_Fiber optic`) en la cúspide de relevancia predictiva, corrigiendo el sesgo de cardinalidad de MDI.

### 2. Formulación económica (hipotética)
Todas las cifras son **supuestos hipotéticos**. Recall y precisión son del modelo campeón en el conjunto de prueba. El simulador de la aplicación recalcula todo con los valores del modelo.

**Supuestos:** base de 50,000 clientes, churn de 26.5%, costo de $35 por contacto, 42% de éxito de la oferta, ARPU de $65/mes, licencia SaaS de $60,000/año.

**Ejemplo** con el campeón de la ejecución de verificación (recall 0.842, precisión 0.480):
1. Churners: 50,000 × 26.5% = 13,250
2. Detectados: 13,250 × 0.842 = 11,159
3. Falsas alarmas: 11,159 × (1 − 0.480) / 0.480 = 12,113
4. Clientes contactados: 11,159 + 12,113 = 23,272. Costo: 23,272 × $35 = $814,520
5. Clientes salvados: 11,159 × 42% = 4,686
6. Ingresos protegidos: 4,686 × $65 × 12 = $3,655,080
7. Inversión: $814,520 + $60,000 = $874,520
8. Beneficio neto: $3,655,080 − $874,520 = **$2,780,560**. No incluye el ahorro de CAC ($1,499,520), por prudencia.
9. ROI: $2,780,560 / $874,520 = **318%**

**Lectura:** el resultado depende sobre todo de la tasa de éxito de la oferta y de las falsas alarmas que implica la precisión del modelo. Si el campeón cambia al repetir la ejecución, estas cifras cambian.
